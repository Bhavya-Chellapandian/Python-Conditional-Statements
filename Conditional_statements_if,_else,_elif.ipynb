Conditional statements - if, else, elif

 Check if a number is positive

```python
num = int(input("Enter the Number"))

if num >= 0 :
  print("Positive Number")
else :
  print("Negative Number")
```

Check if a number is even or odd

```python
n = int(input("Enter the Number"))

if n % 2 == 0:
  print("Even Number")
else :
  print("Odd Number")
```

Check if a number is positive, negative, or zero

```python
num = int(input("Enter the Number"))

if num > 0 :
  print("Positive")
elif num < 0 :
  print("Negative")
else :
  print("Zero")
```

Determine the largest of three numbers

```python
num1 = int(input("Enter the Number"))
num2 = int(input("Enter the Number"))
num3 = int(input("Enter the Number"))

if num1 > num2 and num1 > num3 :
  print("num1 is Largest")
elif num2 > num1 and num2 > num3 :
  print("num2 is Largest")
elif num3 > num1 and num3 > num1 :
  print("num3 is Largest")
else :
  print("Give Three Different Numbers")
```

Print "Pass" or "Fail" based on marks (passing ≥ 35)

```python
Mark = float(input("Enter Your Mark "))

if Mark >= 35 :
  print("Pass")
else :
  print("Fail")
```

Check if a character is a vowel, consonant, or special character

```python
character = input("Enter the Character")

if character.isalpha() :
  if character in "aeiou" :
    print("Vowel")
  else :
    print("Consonant")
else :
  print("Special Character")
```

Assign discount based on purchase amount: ≥1000 → 20%, 500–999 → 10%, <500 → 5%

```python
Purchase_Amount = float(input("Enter the Amount"))

if Purchase_Amount > 0 :
  if Purchase_Amount >= 1000 :
    discount = (Purchase_Amount * 20 ) / 100
    print(f"The Billing Amount is {Purchase_Amount - discount}")

  elif 500 < Purchase_Amount < 1000 :
    discount = (Purchase_Amount * 10 ) / 100
    print(f"The Billing Amount is {Purchase_Amount - discount}")

  else :
    discount = (Purchase_Amount * 5 ) / 100
    print(f"The Billing Amount is {Purchase_Amount - discount }")

else:
  print("No Valid Purchase")
```

Determine ticket price based on age: Child (<12), Adult (12–59), Senior (≥60)

```python
Age = int(input("Enter The Age"))

if Age >= 60 :
  print("Senior Citizen Ticket Fare is 450")

elif 12 <= Age <= 59 :
  print("Adult Ticket Fare is 600")

elif 0 < Age < 12 :
  print("Child Ticket Fare is 300")

else :
  print("Enter Correct Age")
```

Check if a string is uppercase or lowercase

```python
text = str(input("Enter the Text"))

if text.islower():
  print("Text is Lowercase")

elif text.isupper():
  print("Text is Uppercase")

else :
  print("Text in Mixedcase")
```

Classify temperature: Cold (<10°C), Mild (10–25°C), Hot (>25°C)

```python
Temp = float(input("Enter the Temperature"))

if Temp > 25 :
  print("Temperature is Hot")

elif 10 <= Temp <= 25 :
  print("Temperature is Mild")

else :
  print("Temperature is Cold")
```
